```python
def minutes_to_hours_and_minutes():
  """
  This function asks a users how many minutes (they write the number of minutes) they want to be calculate/converted into hours and minutes
  (function converts minutes into hours and remianing minutes into minutes). After the hours and minute results will be printed for user to see
  """
  # first ask user for amount of minutes. - use int(input("...")). See below
  minutes = int(input("How many minutes would you like to to convert into hours and minutes? ")) # This prompt will appear and user will fill out the desired number of minutes.
  # now include the formula for converting minutes into hours (basic division - make sure to use //to get whole numbers).
  hours = minutes // 60
  #now include another formula for the remaining minutes. Use a % to calculate leftover remainder whic are the remaing minute!
  remaining_minutes = minutes % 60

  #Now we need tomake it neat and display the converted hour and minutes nicely by using print and f.
  print(f"Your {minutes} minutes is now converted into {hours} hours and {remaining_minutes} minutes.")

#name of funtion so program can run:
minutes_to_hours_and_minutes()
```
