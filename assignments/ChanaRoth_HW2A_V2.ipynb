```python
def minutes_to_hours_and_minutes():
  """
  This function asks a user how many minutes (they write the number of minutes) they want to be calculate/converted into hours and minutes
  (function converts minutes into hours and remianing minutes into minutes). After the hours and minute results will be printed for user to see
  """
  # first ask user for amount of minutes. - use int(input("...")). See below
  minutes = int(input("How many minutes would you like to convert into hours and minutes? ")) # This prompt will appear and user will fill out the desired number of minutes.
  hours = minutes // 60 # now include the formula for converting minutes into hours (basic division - make sure to use //to get whole numbers).
  remaining_minutes = minutes % 60  #now include another formula for the remaining minutes. Use a % to calculate leftover remainder whic are the remaing minute!
  if hours == 1: #added a conditional statement (an if and else) so grammar will be correct in answer (ex: 1 minute vs seeing 1 minutes)
   shours = str('hour') #here we write str bec it is letters stringed togther to create a word that needs to appear, not an integer.
  else:
    shours = str('hours') #write else so program knows the other option to print when the hour or minute (below) is 1.
  if remaining_minutes == 1:
    sminutes = str('minute')
  else:
    sminutes = str('minutes')


  #Now we need to make it neat and display the converted hours and minutes nicely by using print and f.
  print(f"Your {minutes} minutes is now converted into {hours} {shours} and {remaining_minutes} {sminutes}.")

#name of function so program can run:
minutes_to_hours_and_minutes()
```
